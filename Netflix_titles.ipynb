```python
import pandas as pd
import matplotlib.pyplot as plt

df_netflix = pd.read_csv(r'C:\Users\Admin\PycharmProjects\itproger\netflix_titles.csv')
df_netflix['director'] = df_netflix['director'].fillna('Unknown')
df_netflix['cast'] = df_netflix['cast'].fillna('Unknown')
df_netflix['country'] = df_netflix['country'].fillna('Unknown')
df_netflix['rating'] = df_netflix['rating'].fillna('Unknown')
df_netflix['duration'] = df_netflix['duration'].fillna('Unknown')
df_netflix['date_added'] = df_netflix['date_added'].fillna('Unknown')
```

```python
# ==========================================
# ЗАПИТАННЯ 1: Який рейтинг зустрічається найчастіше?
# ==========================================
# .value_counts() рахує кількість кожного рейтингу, а .idxmax() знаходить лідера
most_common_rating = df_netflix['rating'].value_counts().idxmax()
rating_counts = df_netflix['rating'].value_counts().head(5)

print("=== ВІДПОВІДЬ 1: ТОП-5 РЕЙТИНГІВ ===")
print(rating_counts)
print(f"\n👉 Найчастіше зустрічається рейтинг: {most_common_rating}\n")
```

```python
# ==========================================
# ЗАПИТАННЯ 2: Скільки контенту додали у кожному місяці?
# ==========================================
# Спершу очистимо стовпчик від зайвих пробілів і перетворимо його на формат дати Python
df_netflix['date_added'] = df_netflix['date_added'].str.strip()
df_netflix['date_added'] = pd.to_datetime(df_netflix['date_added'], format='%B %d, %Y', errors='coerce')

# Витягуємо назву місяця
df_netflix['month_added'] = df_netflix['date_added'].dt.strftime('%B')

# Рахуємо кількість для кожного місяця (сортуємо за спаданням кількості контенту)
monthly_content = df_netflix['month_added'].value_counts()

print("=== ВІДПОВІДЬ 2: КОНТЕНТ ЗА МІСЯЦЯМИ ===")
print(monthly_content)
print("\n")
```

```python
# ==========================================
# ЗАПИТАННЯ 3: Які жанри найпопулярніші?
# ==========================================
# У стовпчику listed_in жанри записані через кому (наприклад: "Dramas, International Movies")
# Нам потрібно розділити їх, щоб порахувати кожен жанр окремо
all_genres = df_netflix['listed_in'].str.split(', ').dropna()

# Збираємо всі жанри в один великий плаский список Python
flat_genres = [genre for sublist in all_genres for genre in sublist]

# Перетворюємо назад у Series, щоб легко порахувати топ
top_genres = pd.Series(flat_genres).value_counts().head(10)

print("=== ВІДПОВІДЬ 3: ТОП-10 НАЙПОПУЛЯРНІШИХ ЖАНРІВ ===")
print(top_genres)
```

```python
# ==========================================
# ВІЗУАЛІЗАЦІЯ: Побудуємо графік для жанрів
# ==========================================
top_genres.plot(kind='barh', figsize=(10, 6), color='salmon')
plt.title('ТОП-10 найпопулярніших жанрів на Netflix')
plt.xlabel('Кількість')
plt.tight_layout()
plt.show()
```
